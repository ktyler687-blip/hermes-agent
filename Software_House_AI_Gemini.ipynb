# Software House AI - Gemini Integration

This notebook demonstrates how to connect to the Google Gemini API to analyze our `Software_House_AI.md` system design document.

```python
import google.generativeai as genai
import os

# Configure your API key here (or preferably through an environment variable)
# api_key = os.environ.get("GEMINI_API_KEY")
api_key = "YOUR_GEMINI_API_KEY_HERE"

if api_key != "YOUR_GEMINI_API_KEY_HERE":
    genai.configure(api_key=api_key)
    print("Gemini API configured successfully.")
else:
    print("Please replace 'YOUR_GEMINI_API_KEY_HERE' with your actual Gemini API key.")
```

```python
# Read the Markdown document
try:
    with open('Software_House_AI.md', 'r') as file:
        document_content = file.read()
    print(f"Successfully read {len(document_content)} characters from Software_House_AI.md")
except FileNotFoundError:
    print("Error: Software_House_AI.md not found in the current directory.")
```

```python
# Function to generate a summary using Gemini
def generate_summary():
    if api_key == "YOUR_GEMINI_API_KEY_HERE":
        return "API key not configured. Skipping API call."
        
    model = genai.GenerativeModel('gemini-pro')
    
    prompt = f"""
    Please read the following system design document for a 'Software House AI' 
    and provide a concise, 3-bullet-point summary of its Core Components.
    
    Document:
    {document_content}
    """
    
    print("Sending request to Gemini API...")
    response = model.generate_content(prompt)
    return response.text

# Execute and print the result
print(generate_summary())
```
